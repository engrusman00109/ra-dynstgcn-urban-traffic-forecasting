  <h2 style="color: #92400e; margin-top: 0; font-size: 22px;">2. Problem Statement &amp; Research Motivation</h2>
  <p style="color: #1f2937; font-size: 14.5px; line-height: 1.75;">
    Modern urban transportation networks generate high-frequency, multi-sensor traffic streams from inductive loop detectors, roadside units, and connected vehicles. Classical statistical forecasters and purely recurrent neural networks struggle to capture the non-linear spatial propagation of congestion shockwaves across interconnected freeway segments alongside abrupt temporal regime shifts caused by rush-hour bottlenecks, accidents, or weather anomalies. Addressing this gap requires a unified spatio-temporal graph learning framework that dynamically adapts edge correlations while maintaining real-time edge inference efficiency:
  </p>
  <ul style="color: #374151; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
    <li><span style="background: #fef3c7; color: #92400e; padding: 2px 8px; border-radius: 4px; font-weight: 700;">1. Static Adjacency Bottleneck:</span> Fixed physical road graphs fail to reflect time-varying traffic correlations during morning/evening rush-hour transitions and upstream congestion spillbacks.</li>
    <li><span style="background: #fef3c7; color: #92400e; padding: 2px 8px; border-radius: 4px; font-weight: 700;">2. Long-Horizon Error Accumulation:</span> Multi-step forecasting up to <strong>60 minutes ahead</strong> suffers from severe error drift when roadside loop detectors experience intermittent hardware dropouts (<strong>8.11% missing zero-readings</strong> in METR-LA).</li>
    <li><span style="background: #fef3c7; color: #92400e; padding: 2px 8px; border-radius: 4px; font-weight: 700;">3. Sub-Second IDAS Latency Constraints:</span> On-board and roadside intelligent driving assistance modules demand lightweight models capable of sub-second inference (&lt;1,000 ms) without sacrificing multi-horizon predictive accuracy.</li>
    <li><span style="background: #fef3c7; color: #92400e; padding: 2px 8px; border-radius: 4px; font-weight: 700;">4. Multi-Modal Spatial Heterogeneity:</span> Unifying physical road connectivity, geodesic distance decay, and latent Node2Vec topological embeddings is essential for robust cross-network generalization.</li>
  </ul>

  <h2 style="color: #14532d; margin-top: 0; font-size: 22px;">3. Mathematical Formulation &amp; LaTeX System Architecture Diagrams</h2>

  <p style="color: #1e293b; font-size: 14.5px; line-height: 1.75;">
    Given a historical spatio-temporal tensor $\mathbf{X}_{t-T_{\text{in}}+1:t} \in \mathbb{R}^{T_{\text{in}} \times N \times C}$ across $N = 207$ freeway sensors ($T_{\text{in}} = 12$ steps = $60\text{ min}$, $C = 4$ channels: Z-score normalized speed, $\sin(2\pi\cdot\text{ToD})$, $\cos(2\pi\cdot\text{ToD})$, and $\sin(2\pi\cdot\text{DoW})$), our goal is to learn a non-linear spatio-temporal operator $\mathcal{F}_\Theta(\cdot)$ predicting multi-horizon traffic speeds $\hat{\mathbf{Y}}_{t+1:t+T_{\text{out}}} \in \mathbb{R}^{T_{\text{out}} \times N}$ at $15\text{-min}$ ($h=3$), $30\text{-min}$ ($h=6$), and $60\text{-min}$ ($h=12$) horizons.
  </p>

  <h3 style="color: #166534; font-size: 16px; margin-bottom: 6px;">3.1 Core Mathematical Equations</h3>

  <p style="color: #334155; font-size: 14px; margin-bottom: 4px;"><strong>(a) Multi-Modal Topology Fusion (Dataset 02):</strong></p>
  $$\mathbf{A}_{\text{hybrid}} = 0.55\,\mathbf{A}_{\text{phys}} + 0.25\,\exp\!\left(-\frac{\mathbf{D}_{\text{geo}}^2}{\sigma^2}\right) + 0.20\left(\frac{\mathbf{E}_{\text{SE}}\mathbf{E}_{\text{SE}}^\top}{\|\mathbf{E}_{\text{SE}}\|_2^2} \odot \mathbb{I}[\mathbf{A}_{\text{phys}} > 0]\right)$$

  <p style="color: #334155; font-size: 14px; margin-bottom: 4px;"><strong>(b) Regime-Aware Dynamic Hybrid Graph Constructor (<code>RA-DynSTGCN</code>):</strong></p>
  $$\mathbf{A}_{\text{dyn}}^{(b)} = w_1 \underbrace{\text{Softmax}\!\left(\text{ReLU}(\mathbf{E}_1 \mathbf{E}_2^\top)\right)}_{\text{Adaptive Latent Graph}} + w_2 \underbrace{\text{Softmax}\!\left(\frac{(\mathbf{H}_0^{(b)}\mathbf{W}_q)(\mathbf{H}_0^{(b)}\mathbf{W}_q)^\top}{\sqrt{d}} \odot \mathbf{M}_{\text{supp}}\right)}_{\text{Real-Time Traffic-Regime Edge Attention}} + w_3 \mathbf{P}_f$$

  <p style="color: #334155; font-size: 14px; margin-bottom: 4px;"><strong>(c) Gated Dilated TCN + Bidirectional &amp; Dynamic Diffusion Graph Convolution:</strong></p>
  $$\mathbf{Z}_{\text{TCN}}^{(\ell)} = \tanh\!\left(\mathbf{W}_f^{(\ell)} \star_d \mathbf{H}^{(\ell-1)}\right) \odot \sigma\!\left(\mathbf{W}_g^{(\ell)} \star_d \mathbf{H}^{(\ell-1)}\right)$$
  $$\mathbf{H}^{(\ell)} = \text{BatchNorm}\!\left(\mathbf{H}^{(\ell-1)} + \sum_{k=0}^{K} \left(\mathbf{P}_f^k \mathbf{Z}_{\text{TCN}}^{(\ell)} \mathbf{W}_{f,k} + \mathbf{P}_b^k \mathbf{Z}_{\text{TCN}}^{(\ell)} \mathbf{W}_{b,k}\right) + \mathbf{A}_{\text{dyn}}^{(b)} \mathbf{Z}_{\text{TCN}}^{(\ell)} \mathbf{W}_d\right)$$

  <p style="color: #334155; font-size: 14px; margin-bottom: 4px;"><strong>(d) Horizon-Adaptive Ensemble Gating &amp; Huber Training Objective:</strong></p>
  $$\hat{\mathbf{Y}}_{\text{ens}} = \sum_{m=1}^{3} \text{Softmax}\!\left(\mathbf{G}_{h,n} + f_{\text{ctx}}(\mathbf{x}_t)\right)_m \odot \hat{\mathbf{Y}}^{(m)}, \qquad \mathcal{L}_\delta(y, \hat{y}) = \begin{cases} \frac{1}{2}(y - \hat{y})^2, &amp; |y - \hat{y}| \le \delta \\ \delta\left(|y - \hat{y}| - \frac{1}{2}\delta\right), &amp; |y - \hat{y}| &gt; \delta \end{cases}$$

  <h2 style="color: #1e40af; margin-top: 0; font-size: 21px;">Step 1: Environment Initialization, Multi-Modal Graph Construction &amp; Exploratory Data Analysis</h2>
  <p style="color: #1e293b; font-size: 14px; line-height: 1.65;">
    In this foundational stage, we initialize the PyTorch GPU/CPU compute environment, extract both attached dataset archives (<code>Dataset01.zip</code> and <code>Dataset02.zip</code>), repair sensor dropouts, and inspect diurnal traffic dynamics:
  </p>
  <ul style="color: #1e293b; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
    <li><strong>Dual-Archive Ingestion &amp; Multi-Modal Schema Parsing:</strong> Extracts <code>METR-LA.h5</code> and <code>adj_METR-LA.pkl</code> from <code>Dataset01.zip</code>, alongside <code>metr-la.h5</code>, <code>adj_mx.pkl</code>, 64-D Node2Vec embeddings (<code>SE_metrla.txt</code>), pairwise road distances (<code>distances_la_2012.csv</code>), and GPS coordinates (<code>graph_sensor_locations.csv</code>) from <code>Dataset02.zip</code>.</li>
    <li><strong>Two-Stage Sensor Dropout Imputation &amp; Graph Smoothing:</strong> Replaces exact <code>0.0 mph</code> hardware dropouts (<strong>8.11% missing rate</strong>) via per-sensor temporal linear interpolation, followed by a 10% spatial neighborhood weighted mean blend ($\mathbf{X}_{\text{imp}} \leftarrow 0.90\mathbf{X}_{\text{imp}} + 0.10\tilde{\mathbf{A}}\mathbf{X}_{\text{imp}}$) to suppress boundary artifacts.</li>
    <li><strong>Hybrid Multi-Modal Adjacency Fusion (Dataset 02):</strong> Combines physical connectivity ($\mathbf{A}_{\text{phys}}$), geodesic Gaussian distance kernels ($\exp(-d_{ij}^2/\sigma^2)$), and Node2Vec embedding cosine similarity into a unified spatial operator $\mathbf{A}_{\text{hybrid}}$.</li>
    <li><strong>Cyclical Temporal Covariate Encoding:</strong> Appends three periodic time channels—$\sin(2\pi\cdot\text{ToD})$, $\cos(2\pi\cdot\text{ToD})$, and $\sin(2\pi\cdot\text{DoW})$—yielding a rich 4-channel spatio-temporal tensor $\mathbf{X} \in \mathbb{R}^{12 \times 207 \times 4}$ per 60-minute input window.</li>
    <li><strong>Chronological Partitioning &amp; Leakage-Free Normalization:</strong> Splits all 34,272 timestamps chronologically into <strong>70% Train</strong>, <strong>10% Validation</strong>, and <strong>20% Test</strong> subsets, fitting Z-score parameters ($\mu = 58.36\text{ mph}, \sigma = 12.98\text{ mph}$) exclusively on the training block.</li>

```python
# ==============================================================================
# STEP 1A: ENVIRONMENT SETUP, REPRODUCIBILITY & EXPERIMENT CONFIGURATION
# ==============================================================================
import os
import zipfile
import pickle
import gc
import time
import copy
import json
import shutil
from typing import Dict, Tuple, Optional, List

# Data Science & Machine Learning Stack
import numpy as np
import pandas as pd
import h5py
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Visualization Stack
import matplotlib.pyplot as plt
import seaborn as sns

# Evaluation Metrics
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score, confusion_matrix

# Reproducibility
def seed_everything(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

seed_everything(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Create output directories
for path in ["outputs/figures", "outputs/checkpoints", "outputs/tables", "outputs/onnx"]:
    os.makedirs(path, exist_ok=True)

# Experiment Configuration
class ExperimentConfig:
    def __init__(self):
        self.num_nodes = 207
        self.seq_len = 12
        self.pred_len = 12
        self.train_ratio = 0.7
        self.val_ratio = 0.1
        self.train_stride = 6
        self.eval_stride = 12
        self.batch_size = 64
        self.epochs = 5  # Kept small for fast Colab demo execution
        self.lr = 3e-3
        self.weight_decay = 1e-4
        self.huber_delta = 1.0
        self.dropout = 0.15
        self.hidden_dim = 28
        self.embed_dim = 16
        self.diffusion_steps = 2
        self.congestion_threshold_mph = 45.0
        self.eval_horizons = (2, 5, 11)  # Represents 15-min, 30-min, and 60-min indexes

cfg = ExperimentConfig()
print("Configuration and imports successfully initialized.")
```

```python
# ==============================================================================
# STEP 1B: UNIFIED MULTI-MODAL GRAPH DATA LOADER & EXPLORATORY DATA ANALYSIS
# ==============================================================================
class SpatioTemporalGraphLoader:
    """
    Unified dataset acquisition, spatial graph fusion, missing-value imputation,
    and chronological window generator for Dataset01 and Dataset02.
    """

    def __init__(self, extract_root: str = "/tmp/stgnn_colab_extract"):
        self.extract_root = extract_root
        os.makedirs(self.extract_root, exist_ok=True)

    @staticmethod
    def _impute_and_smooth(speed_raw: np.ndarray, adj: np.ndarray) -> np.ndarray:
        orig_zero_mask = (speed_raw == 0.0)
        df = pd.DataFrame(speed_raw)
        df.replace(0.0, np.nan, inplace=True)
        imputed = df.interpolate(method="linear", axis=0, limit_direction="both").values.astype(np.float32)
        del df

        row_sums = adj.sum(axis=1, keepdims=True)
        norm_adj = np.divide(adj, np.maximum(row_sums, 1e-6))
        nan_mask = np.isnan(imputed)
        if nan_mask.any():
            filled_temp = np.nan_to_num(imputed, nan=float(np.nanmean(imputed)))
            neighbor_mean = filled_temp @ norm_adj.T
            imputed[nan_mask] = neighbor_mean[nan_mask]

        neighbor_all = imputed @ norm_adj.T
        imputed[orig_zero_mask] = 0.90 * imputed[orig_zero_mask] + 0.10 * neighbor_all[orig_zero_mask]
        return imputed.astype(np.float32)

    @staticmethod
    def _compute_time_features(timestamps: np.ndarray) -> np.ndarray:
        dt = pd.to_datetime(timestamps)
        tod = (dt.hour * 60 + dt.minute) / (24.0 * 60.0)
        dow = dt.dayofweek / 7.0
        return np.stack([
            np.sin(2 * np.pi * tod).astype(np.float32),
            np.cos(2 * np.pi * tod).astype(np.float32),
            np.sin(2 * np.pi * dow).astype(np.float32),
        ], axis=-1)

    def load_dataset_01(self, zip_path: str) -> Dict:
        target_dir = os.path.join(self.extract_root, "dataset01")
        with zipfile.ZipFile(zip_path, "r") as zf:
            zf.extractall(target_dir)

        with h5py.File(os.path.join(target_dir, "METR-LA.h5"), "r") as f:
            raw_speed = f["df"]["block0_values"][:].astype(np.float32)
            timestamps = f["df"]["axis1"][:]
            sensor_ids = [s.decode("utf-8") if isinstance(s, bytes) else str(s) for s in f["df"]["axis0"][:]]

        with open(os.path.join(target_dir, "adj_METR-LA.pkl"), "rb") as f:
            _, _, adj_mx = pickle.load(f, encoding="latin1")
        adj_mx = adj_mx.astype(np.float32)
        np.fill_diagonal(adj_mx, 1.0)

        zero_rate = float((raw_speed == 0.0).mean())
        clean_speed = self._impute_and_smooth(raw_speed, adj_mx)
        del raw_speed
        gc.collect()
        shutil.rmtree(target_dir, ignore_errors=True)

        return {
            "name": "Dataset01_Physical_RoadGraph",
            "speed": clean_speed,
            "adj": adj_mx,
            "time_feats": self._compute_time_features(timestamps),
            "timestamps": timestamps,
            "sensor_ids": sensor_ids,
            "spatial_emb": None,
            "coords": None,
            "zero_rate": zero_rate,
        }

    def load_dataset_02(self, zip_path: str) -> Dict:
        target_dir = os.path.join(self.extract_root, "dataset02")
        with zipfile.ZipFile(zip_path, "r") as zf:
            zf.extractall(target_dir)

        with h5py.File(os.path.join(target_dir, "metr-la.h5"), "r") as f:
            raw_speed = f["df"]["block0_values"][:].astype(np.float32)
            timestamps = f["df"]["axis1"][:]
            sensor_ids = [s.decode("utf-8") if isinstance(s, bytes) else str(s) for s in f["df"]["axis0"][:]]

        with open(os.path.join(target_dir, "adj_mx.pkl"), "rb") as f:
            _, id_to_idx, adj_mx = pickle.load(f, encoding="latin1")
        adj_mx = adj_mx.astype(np.float32)

        with open(os.path.join(target_dir, "SE_metrla.txt"), "r") as f:
            header = f.readline().strip().split()
            se_dim = int(header[1])
            se_matrix = np.zeros((len(sensor_ids), se_dim), dtype=np.float32)
            for line in f:
                parts = line.strip().split()
                if parts and int(parts[0]) < len(sensor_ids):
                    se_matrix[int(parts[0])] = np.array([float(v) for v in parts[1:]], dtype=np.float32)

        se_norm = se_matrix / np.maximum(np.linalg.norm(se_matrix, axis=1, keepdims=True), 1e-6)
        se_sim = np.clip(se_norm @ se_norm.T, 0.0, 1.0)

        df_dist = pd.read_csv(os.path.join(target_dir, "distances_la_2012.csv"))
        dist_mat = np.full((len(sensor_ids), len(sensor_ids)), np.inf, dtype=np.float32)
        np.fill_diagonal(dist_mat, 0.0)
        for row in df_dist.itertuples(index=False):
            u, v, c = str(row[0]), str(row[1]), float(row[2])
            if u in id_to_idx and v in id_to_idx:
                dist_mat[id_to_idx[u], id_to_idx[v]] = c
        del df_dist
        finite_dists = dist_mat[np.isfinite(dist_mat) & (dist_mat > 0)]
        sigma = float(np.std(finite_dists)) if len(finite_dists) > 0 else 1000.0
        geo_adj = np.exp(-np.square(dist_mat / max(sigma, 1.0)))
        geo_adj[geo_adj < 0.1] = 0.0

        hybrid_adj = 0.55 * adj_mx + 0.25 * geo_adj + 0.20 * (se_sim * (adj_mx > 0).astype(np.float32))
        np.fill_diagonal(hybrid_adj, 1.0)

        df_loc = pd.read_csv(os.path.join(target_dir, "graph_sensor_locations.csv")).sort_values("index")
        coords = df_loc[["latitude", "longitude"]].values.astype(np.float32)

        zero_rate = float((raw_speed == 0.0).mean())
        clean_speed = self._impute_and_smooth(raw_speed, hybrid_adj)
        del raw_speed
        gc.collect()
        shutil.rmtree(target_dir, ignore_errors=True)

        return {
            "name": "Dataset02_MultiModal_SpatialTopology",
            "speed": clean_speed,
            "adj": hybrid_adj.astype(np.float32),
            "time_feats": self._compute_time_features(timestamps),
            "timestamps": timestamps,
            "sensor_ids": sensor_ids,
            "spatial_emb": se_matrix,
            "coords": coords,
            "zero_rate": zero_rate,
        }


class TrafficWindowDataset(Dataset):
    def __init__(self, speed: np.ndarray, time_feats: np.ndarray, mean: float, std: float,
                 seq_len: int = 12, pred_len: int = 12, stride: int = 6):
        self.speed_norm = ((speed - mean) / std).astype(np.float32)
        self.time_feats = time_feats.astype(np.float32)
        self.seq_len = seq_len
        self.pred_len = pred_len
        max_start = len(speed) - seq_len - pred_len + 1
        self.indices = np.arange(0, max_start, stride, dtype=np.int64)

    def __len__(self) -> int:
        return len(self.indices)

    def __getitem__(self, idx: int) -> Tuple[torch.Tensor, torch.Tensor]:
        s = self.indices[idx]
        x_speed = self.speed_norm[s : s + self.seq_len]
        t_feat = self.time_feats[s : s + self.seq_len]
        N = x_speed.shape[1]
        t_broad = np.broadcast_to(t_feat[:, None, :], (self.seq_len, N, 3))
        x = np.concatenate([x_speed[..., None], t_broad], axis=-1)
        y = self.speed_norm[s + self.seq_len : s + self.seq_len + self.pred_len]
        return torch.from_numpy(x.copy()), torch.from_numpy(y.copy())


def create_dataloaders(data_dict: Dict, cfg: ExperimentConfig):
    speed, time_feats = data_dict["speed"], data_dict["time_feats"]
    T = len(speed)
    n_train = int(T * cfg.train_ratio)
    n_val = int(T * (cfg.train_ratio + cfg.val_ratio))

    mean, std = float(speed[:n_train].mean()), float(speed[:n_train].std())
    train_ds = TrafficWindowDataset(speed[:n_train], time_feats[:n_train], mean, std, cfg.seq_len, cfg.pred_len, cfg.train_stride)
    val_ds = TrafficWindowDataset(speed[n_train:n_val], time_feats[n_train:n_val], mean, std, cfg.seq_len, cfg.pred_len, cfg.eval_stride)
    test_ds = TrafficWindowDataset(speed[n_val:], time_feats[n_val:], mean, std, cfg.seq_len, cfg.pred_len, cfg.eval_stride)

    return (
        DataLoader(train_ds, batch_size=cfg.batch_size, shuffle=True),
        DataLoader(val_ds, batch_size=cfg.batch_size, shuffle=False),
        DataLoader(test_ds, batch_size=cfg.batch_size, shuffle=False),
        {"mean": mean, "std": std},
    )

# Load both datasets and display summary audit + EDA charts
loader = SpatioTemporalGraphLoader()
ds1 = loader.load_dataset_01("Dataset01.zip")
ds2 = loader.load_dataset_02("Dataset02.zip")

eda_df = pd.DataFrame([
    {
        "Dataset": ds1["name"],
        "Time_Steps": ds1["speed"].shape[0],
        "Sensors": ds1["speed"].shape[1],
        "Raw_Zero_Dropout_%": round(ds1["zero_rate"] * 100, 2),
        "Mean_Speed_mph": round(float(ds1["speed"].mean()), 2),
        "Std_Speed_mph": round(float(ds1["speed"].std()), 2),
        "Graph_Density_%": round(float((ds1["adj"] > 0).mean() * 100), 2),
        "Extra_Modalities": "Physical Adjacency Only",
    },
    {
        "Dataset": ds2["name"],
        "Time_Steps": ds2["speed"].shape[0],
        "Sensors": ds2["speed"].shape[1],
        "Raw_Zero_Dropout_%": round(ds2["zero_rate"] * 100, 2),
        "Mean_Speed_mph": round(float(ds2["speed"].mean()), 2),
        "Std_Speed_mph": round(float(ds2["speed"].std()), 2),
        "Graph_Density_%": round(float((ds2["adj"] > 0).mean() * 100), 2),
        "Extra_Modalities": "Node2Vec (64D) + Geodesic Distances + GPS",
    }
])
display(eda_df)

# Exploratory Visualizations: Diurnal Speed Profile, Speed Distribution & Spatial Graph Comparison
plt.style.use("seaborn-v0_8-whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))

# Plot 1: Diurnal Rush-Hour Profile across 24 Hours
dt_idx = pd.to_datetime(ds2["timestamps"])
hourly_speed = pd.DataFrame({"hour": dt_idx.hour, "speed": ds2["speed"].mean(axis=1)}).groupby("hour")["speed"].agg(["mean", "std"])
axes[0].plot(hourly_speed.index, hourly_speed["mean"], "o-", color="#1d4ed8", linewidth=2.2, label="Network Mean Speed")
axes[0].fill_between(hourly_speed.index, hourly_speed["mean"] - hourly_speed["std"], hourly_speed["mean"] + hourly_speed["std"], color="#93c5fd", alpha=0.35, label="±1 Std Dev")
axes[0].axhline(cfg.congestion_threshold_mph, color="#dc2626", linestyle="--", linewidth=1.3, label=f"Congestion Threshold ({cfg.congestion_threshold_mph} mph)")
axes[0].set_title("Diurnal Urban Speed Profile (Morning/Evening Peaks)", fontsize=11, fontweight="bold")
axes[0].set_xlabel("Hour of Day (0 - 23)")
axes[0].set_ylabel("Traffic Speed (mph)")
axes[0].legend(fontsize=8)

# Plot 2: Imputed Speed Distribution
axes[1].hist(ds2["speed"][::10].ravel(), bins=45, color="#0284c7", edgecolor="white", alpha=0.85)
axes[1].axvline(cfg.congestion_threshold_mph, color="#dc2626", linestyle="--", linewidth=1.5, label="Severe Congestion Cutoff")
axes[1].set_title("Imputed Network Speed Distribution (METR-LA)", fontsize=11, fontweight="bold")
axes[1].set_xlabel("Traffic Speed (mph)")
axes[1].set_ylabel("Frequency")
axes[1].legend(fontsize=8)

# Plot 3: Dataset02 Fused Multi-Modal Adjacency Subgraph
im = axes[2].imshow(ds2["adj"][:45, :45], cmap="magma", aspect="auto")
axes[2].set_title("Dataset 02 Hybrid Fused Adjacency (45x45 Subgraph)", fontsize=11, fontweight="bold")
axes[2].set_xlabel("Target Sensor Index")
axes[2].set_ylabel("Source Sensor Index")
plt.colorbar(im, ax=axes[2], label="Fused Edge Weight")

plt.tight_layout()
plt.savefig("outputs/figures/step1_eda_overview.png", dpi=180)
plt.show()
```

```python
# ==============================================================================
# STEP 2: 3 BUILT-IN MODELS, CUSTOM RA-DynSTGCN (FROM SCRATCH) & GATED ENSEMBLE
# ==============================================================================

# --- 1. Built-In Model 1: LSTM ---
class BuiltInLSTMForecaster(nn.Module):
    def __init__(self, in_channels: int = 4, hidden_dim: int = 36, pred_len: int = 12, dropout: float = 0.15):
        super().__init__()
        self.pred_len = pred_len
        self.lstm = nn.LSTM(in_channels, hidden_dim, num_layers=2, batch_first=True, dropout=dropout)
        self.head = nn.Sequential(nn.Linear(hidden_dim, hidden_dim), nn.GELU(), nn.Linear(hidden_dim, pred_len))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, N, C = x.shape
        out, _ = self.lstm(x.permute(0, 2, 1, 3).reshape(B * N, T, C))
        return self.head(out[:, -1, :]).view(B, N, self.pred_len).permute(0, 2, 1)


# --- 2. Built-In Model 2: GRU + Temporal Attention ---
class BuiltInGRUForecaster(nn.Module):
    def __init__(self, in_channels: int = 4, hidden_dim: int = 36, pred_len: int = 12, dropout: float = 0.15):
        super().__init__()
        self.pred_len = pred_len
        self.gru = nn.GRU(in_channels, hidden_dim, num_layers=2, batch_first=True, dropout=dropout)
        self.attn = nn.Linear(hidden_dim, 1)
        self.head = nn.Sequential(nn.Linear(hidden_dim * 2, hidden_dim), nn.SiLU(), nn.Linear(hidden_dim, pred_len))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, N, C = x.shape
        out, _ = self.gru(x.permute(0, 2, 1, 3).reshape(B * N, T, C))
        weights = torch.softmax(self.attn(out), dim=1)
        ctx = torch.sum(out * weights, dim=1)
        pred = self.head(torch.cat([out[:, -1, :], ctx], dim=-1))
        return pred.view(B, N, self.pred_len).permute(0, 2, 1)


# --- 3. Built-In Model 3: Spatio-Temporal TransformerEncoder ---
class BuiltInTransformerForecaster(nn.Module):
    def __init__(self, num_nodes: int = 207, in_channels: int = 4, d_model: int = 24,
                 nhead: int = 4, pred_len: int = 12, dropout: float = 0.15):
        super().__init__()
        self.pred_len = pred_len
        self.input_proj = nn.Linear(in_channels, d_model)
        self.node_emb = nn.Parameter(torch.randn(1, 1, num_nodes, d_model) * 0.02)
        self.pos_emb = nn.Parameter(torch.randn(1, 12, 1, d_model) * 0.02)
        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=d_model * 2,
            dropout=dropout, activation="gelu", batch_first=True
        )
        self.encoder = nn.TransformerEncoder(enc_layer, num_layers=2)
        self.head = nn.Linear(d_model, pred_len)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, N, C = x.shape
        h = self.input_proj(x) + self.node_emb[:, :, :N, :] + self.pos_emb[:, :T, :, :]
        enc = self.encoder(h.permute(0, 2, 1, 3).reshape(B * N, T, -1))
        return self.head(enc[:, -1, :]).view(B, N, self.pred_len).permute(0, 2, 1)


# --- 4. Proposed Custom Research Model from Scratch: RA-DynSTGCN ---
class DynamicHybridGraphConstructor(nn.Module):
    """
    Fuses physical forward/backward random-walk matrices (P_f, P_b),
    learnable adaptive node embeddings (E1, E2), and a traffic-regime-conditioned
    dynamic edge attention gate.
    """
    def __init__(self, adj_mx: np.ndarray, num_nodes: int = 207, embed_dim: int = 16,
                 hidden_dim: int = 28, pretrained_se: Optional[np.ndarray] = None):
        super().__init__()
        adj = adj_mx.astype(np.float32)
        p_fwd = adj / np.maximum(adj.sum(axis=1, keepdims=True), 1e-6)
        p_bwd = adj.T / np.maximum(adj.T.sum(axis=1, keepdims=True), 1e-6)
        self.register_buffer("p_fwd", torch.from_numpy(p_fwd))
        self.register_buffer("p_bwd", torch.from_numpy(p_bwd))

        if pretrained_se is not None:
            se_tensor = torch.from_numpy(pretrained_se[:, :embed_dim].astype(np.float32))
            self.node_vec1 = nn.Parameter(se_tensor.clone())
            self.node_vec2 = nn.Parameter(se_tensor.clone().T)
        else:
            self.node_vec1 = nn.Parameter(torch.randn(num_nodes, embed_dim) * 0.1)
            self.node_vec2 = nn.Parameter(torch.randn(embed_dim, num_nodes) * 0.1)

        self.regime_proj = nn.Linear(hidden_dim, embed_dim)
        self.alpha_logits = nn.Parameter(torch.tensor([0.4, 0.3, 0.3]))

    def forward(self, node_context: torch.Tensor):
        a_adapt = F.softmax(F.relu(torch.mm(self.node_vec1, self.node_vec2)), dim=-1)
        q = self.regime_proj(node_context)
        scores = torch.matmul(q, q.transpose(1, 2)) / (q.shape[-1] ** 0.5)
        support_mask = ((self.p_fwd + self.p_bwd) > 0).float().unsqueeze(0)
        a_state = F.softmax(scores.masked_fill(support_mask == 0, -1e9), dim=-1)

        w = F.softmax(self.alpha_logits, dim=0)
        a_dyn = w[0] * a_adapt.unsqueeze(0) + w[1] * a_state + w[2] * self.p_fwd.unsqueeze(0)
        return self.p_fwd, self.p_bwd, a_dyn


class MultiScaleDiffusionGraphConv(nn.Module):
    def __init__(self, c_in: int, c_out: int, diffusion_steps: int = 2, dropout: float = 0.15):
        super().__init__()
        self.K = diffusion_steps
        num_supports = 1 + 2 * diffusion_steps + 1
        self.mlp = nn.Conv2d(c_in * num_supports, c_out, kernel_size=(1, 1))
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor, p_fwd: torch.Tensor, p_bwd: torch.Tensor, a_dyn: torch.Tensor) -> torch.Tensor:
        supports = [x]
        x_f, x_b = x, x
        for _ in range(self.K):
            x_f = torch.einsum("bcnt,nm->bcmt", x_f, p_fwd)
            x_b = torch.einsum("bcnt,nm->bcmt", x_b, p_bwd)
            supports.extend([x_f, x_b])
        supports.append(torch.einsum("bcnt,bnm->bcmt", x, a_dyn))
        return self.dropout(self.mlp(torch.cat(supports, dim=1)))


class GatedDilatedTCNBlock(nn.Module):
    def __init__(self, hidden_dim: int, kernel_size: int = 3, dilation: int = 1, dropout: float = 0.15):
        super().__init__()
        self.padding = (kernel_size - 1) * dilation
        self.filter_conv = nn.Conv2d(hidden_dim, hidden_dim, kernel_size=(1, kernel_size), dilation=(1, dilation))
        self.gate_conv = nn.Conv2d(hidden_dim, hidden_dim, kernel_size=(1, kernel_size), dilation=(1, dilation))
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x_pad = F.pad(x, (self.padding, 0))
        return self.dropout(torch.tanh(self.filter_conv(x_pad)) * torch.sigmoid(self.gate_conv(x_pad)))


class CustomDynamicSTGCN(nn.Module):
    """Proposed Research Model Built from Scratch: RA-DynSTGCN."""
    def __init__(self, adj_mx: np.ndarray, num_nodes: int = 207, in_channels: int = 4,
                 hidden_dim: int = 28, embed_dim: int = 16, pred_len: int = 12,
                 diffusion_steps: int = 2, dropout: float = 0.15, pretrained_se: Optional[np.ndarray] = None):
        super().__init__()
        self.start_conv = nn.Conv2d(in_channels, hidden_dim, kernel_size=(1, 1))
        self.graph_builder = DynamicHybridGraphConstructor(adj_mx, num_nodes, embed_dim, hidden_dim, pretrained_se)

        self.tcn1 = GatedDilatedTCNBlock(hidden_dim, kernel_size=3, dilation=1, dropout=dropout)
        self.gcn1 = MultiScaleDiffusionGraphConv(hidden_dim, hidden_dim, diffusion_steps, dropout)
        self.norm1 = nn.BatchNorm2d(hidden_dim)

        self.tcn2 = GatedDilatedTCNBlock(hidden_dim, kernel_size=3, dilation=2, dropout=dropout)
        self.gcn2 = MultiScaleDiffusionGraphConv(hidden_dim, hidden_dim, diffusion_steps, dropout)
        self.norm2 = nn.BatchNorm2d(hidden_dim)

        self.skip_conv = nn.Conv2d(hidden_dim * 2, hidden_dim * 2, kernel_size=(1, 1))
        self.end_conv_1 = nn.Conv2d(hidden_dim * 2, hidden_dim * 2, kernel_size=(1, 1))
        self.end_conv_2 = nn.Conv2d(hidden_dim * 2, pred_len, kernel_size=(1, 1))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        h0 = self.start_conv(x.permute(0, 3, 2, 1))
        p_fwd, p_bwd, a_dyn = self.graph_builder(h0.mean(dim=-1).permute(0, 2, 1))

        h1 = self.norm1(h0 + self.gcn1(self.tcn1(h0), p_fwd, p_bwd, a_dyn))
        h2 = self.norm2(h1 + self.gcn2(self.tcn2(h1), p_fwd, p_bwd, a_dyn))

        skip = torch.cat([h1[..., -1:], h2[..., -1:]], dim=1)
        out = F.gelu(self.skip_conv(skip))
        out = F.gelu(self.end_conv_1(out))
        return self.end_conv_2(out).squeeze(-1)

    @torch.no_grad()
    def extract_dynamic_adjacency(self, x: torch.Tensor) -> np.ndarray:
        self.eval()
        h0 = self.start_conv(x.permute(0, 3, 2, 1))
        _, _, a_dyn = self.graph_builder(h0.mean(dim=-1).permute(0, 2, 1))
        return a_dyn.mean(dim=0).cpu().numpy()


# --- 5. Model 5: Horizon-Adaptive Gated Ensemble ---
class HorizonAdaptiveEnsembleForecaster(nn.Module):
    def __init__(self, num_nodes: int = 207, in_channels: int = 4, hidden_dim: int = 36,
                 pred_len: int = 12, dropout: float = 0.15):
        super().__init__()
        self.branch_lstm = BuiltInLSTMForecaster(in_channels, hidden_dim, pred_len, dropout)
        self.branch_gru = BuiltInGRUForecaster(in_channels, hidden_dim, pred_len, dropout)
        self.branch_trans = BuiltInTransformerForecaster(num_nodes, in_channels, 24, 4, pred_len, dropout)

        self.horizon_gate = nn.Parameter(torch.zeros(1, pred_len, num_nodes, 3))
        self.ctx_gate = nn.Sequential(nn.Linear(in_channels, 16), nn.GELU(), nn.Linear(16, 3))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        stacked = torch.stack([self.branch_lstm(x), self.branch_gru(x), self.branch_trans(x)], dim=-1)
        ctx_logits = self.ctx_gate(x[:, -1, :, :]).unsqueeze(1)
        weights = F.softmax(self.horizon_gate + ctx_logits, dim=-1)
        return torch.sum(stacked * weights, dim=-1)


def build_five_models(cfg: ExperimentConfig, adj_mx: np.ndarray, spatial_emb=None):
    return {
        "1_BuiltIn_LSTM": BuiltInLSTMForecaster(4, 36, cfg.pred_len, cfg.dropout),
        "2_BuiltIn_GRU_Attn": BuiltInGRUForecaster(4, 36, cfg.pred_len, cfg.dropout),
        "3_BuiltIn_Transformer": BuiltInTransformerForecaster(cfg.num_nodes, 4, 24, 4, cfg.pred_len, cfg.dropout),
        "4_Proposed_RA_DynSTGCN": CustomDynamicSTGCN(
            adj_mx=adj_mx,
            num_nodes=cfg.num_nodes,
            in_channels=4,
            hidden_dim=cfg.hidden_dim,
            embed_dim=cfg.embed_dim,
            pred_len=cfg.pred_len,
            diffusion_steps=cfg.diffusion_steps,
            dropout=cfg.dropout,
            pretrained_se=spatial_emb,
        ),
        "5_HorizonAdaptive_Ensemble": HorizonAdaptiveEnsembleForecaster(
            cfg.num_nodes, 4, 36, cfg.pred_len, cfg.dropout
        ),
    }

sample_models = build_five_models(cfg, ds2["adj"], ds2["spatial_emb"])
param_summary = pd.DataFrame([
    {"Model": k, "Trainable_Parameters": sum(p.numel() for p in m.parameters() if p.requires_grad)}
    for k, m in sample_models.items()
])
display(param_summary)
del sample_models
```

```python
# ==============================================================================
# STEP 3A: MULTI-HORIZON TRAINING & EVALUATION ACROSS DATASET 01 & DATASET 02
# ==============================================================================
def compute_regression_metrics(preds: np.ndarray, targets: np.ndarray) -> Dict[str, float]:
    diff = preds - targets
    mae = float(np.mean(np.abs(diff)))
    rmse = float(np.sqrt(np.mean(np.square(diff))))
    mask = targets > 5.0
    mape = float(np.mean(np.abs(diff[mask] / targets[mask])) * 100.0) if np.any(mask) else 0.0
    return {"MAE": round(mae, 4), "RMSE": round(rmse, 4), "MAPE": round(mape, 4)}


class MultiHorizonTrainer:
    def __init__(self, model: nn.Module, device: torch.device, cfg: ExperimentConfig):
        self.model = model.to(device)
        self.device = device
        self.epochs = cfg.epochs
        self.criterion = nn.HuberLoss(delta=cfg.huber_delta)
        self.optimizer = torch.optim.AdamW(self.model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
        self.scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(self.optimizer, T_max=max(cfg.epochs, 1), eta_min=1e-5)

    def fit(self, train_loader: DataLoader, val_loader: DataLoader) -> Dict:
        best_val_loss = float("inf")
        best_weights = copy.deepcopy(self.model.state_dict())
        history = {"train_loss": [], "val_loss": []}
        t0 = time.time()

        for epoch in range(1, self.epochs + 1):
            self.model.train()
            tr_losses = []
            for x, y in train_loader:
                x, y = x.to(self.device), y.to(self.device)
                self.optimizer.zero_grad()
                loss = self.criterion(self.model(x), y)
                loss.backward()
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=5.0)
                self.optimizer.step()
                tr_losses.append(loss.item())
            self.scheduler.step()

            self.model.eval()
            va_losses = []
            with torch.no_grad():
                for x, y in val_loader:
                    x, y = x.to(self.device), y.to(self.device)
                    va_losses.append(self.criterion(self.model(x), y).item())

            tr_l, va_l = float(np.mean(tr_losses)), float(np.mean(va_losses))
            history["train_loss"].append(tr_l)
            history["val_loss"].append(va_l)
            if va_l < best_val_loss:
                best_val_loss = va_l
                best_weights = copy.deepcopy(self.model.state_dict())

        self.model.load_state_dict(best_weights)
        history["train_time_sec"] = round(time.time() - t0, 2)
        return history

    @torch.no_grad()
    def evaluate(self, test_loader: DataLoader, scaler: Dict, eval_horizons: Tuple[int, int, int] = (2, 5, 11)):
        self.model.eval()
        preds_list, targets_list = [], []
        mean, std = scaler["mean"], scaler["std"]
        for x, y in test_loader:
            out = self.model(x.to(self.device)).cpu().numpy()
            preds_list.append(out * std + mean)
            targets_list.append(y.numpy() * std + mean)

        preds = np.concatenate(preds_list, axis=0)
        targets = np.concatenate(targets_list, axis=0)
        results = {"Overall": compute_regression_metrics(preds, targets)}
        for h_idx, label in zip(eval_horizons, ["15-min", "30-min", "60-min"]):
            results[label] = compute_regression_metrics(preds[:, h_idx, :], targets[:, h_idx, :])
        return results, preds, targets


all_rows = []
training_histories = {"Dataset01_Physical_RoadGraph": {}, "Dataset02_MultiModal_SpatialTopology": {}}
d2_test_preds = {}
d2_test_targets = None
d2_dyn_adj = None
d2_sample_x = None
trained_d2_stgcn = None

for ds in [ds1, ds2]:
    ds_name = ds["name"]
    print(f"\n{'='*72}\n BENCHMARKING ALL 5 MODELS ON: {ds_name}\n{'='*72}")
    train_loader, val_loader, test_loader, scaler = create_dataloaders(ds, cfg)
    models = build_five_models(cfg, ds["adj"], ds["spatial_emb"])

    for model_name, model in models.items():
        trainer = MultiHorizonTrainer(model, device, cfg)
        hist = trainer.fit(train_loader, val_loader)
        training_histories[ds_name][model_name] = hist
        metrics, preds, targets = trainer.evaluate(test_loader, scaler, cfg.eval_horizons)
        torch.save(model.state_dict(), f"outputs/checkpoints/{ds_name}_{model_name}.pt")

        row = {
            "Dataset": ds_name,
            "Model": model_name,
            "Params": sum(p.numel() for p in model.parameters()),
            "Train_Time_s": hist["train_time_sec"],
            "Overall_MAE": metrics["Overall"]["MAE"],
            "Overall_RMSE": metrics["Overall"]["RMSE"],
            "Overall_MAPE": metrics["Overall"]["MAPE"],
            "15min_MAE": metrics["15-min"]["MAE"],
            "30min_MAE": metrics["30-min"]["MAE"],
            "60min_MAE": metrics["60-min"]["MAE"],
        }
        all_rows.append(row)
        print(f"{model_name:28s} | MAE: {row['Overall_MAE']:.4f} | RMSE: {row['Overall_RMSE']:.4f} | MAPE: {row['Overall_MAPE']:.2f}% | 15m: {row['15min_MAE']:.4f} | 30m: {row['30min_MAE']:.4f} | 60m: {row['60min_MAE']:.4f}")

        if ds_name == "Dataset02_MultiModal_SpatialTopology":
            d2_test_preds[model_name] = preds
            if d2_test_targets is None:
                d2_test_targets = targets
            if model_name == "4_Proposed_RA_DynSTGCN":
                sample_x, _ = next(iter(test_loader))
                d2_sample_x = sample_x
                d2_dyn_adj = model.extract_dynamic_adjacency(sample_x.to(device))
                trained_d2_stgcn = copy.deepcopy(model).cpu()

df_summary = pd.DataFrame(all_rows)
df_summary.to_csv("outputs/tables/cross_dataset_5model_comparison.csv", index=False)

piv_summary = df_summary.pivot(
    index="Model",
    columns="Dataset",
    values=["Overall_MAE", "Overall_RMSE", "Overall_MAPE", "15min_MAE", "30min_MAE", "60min_MAE"]
)
piv_summary.to_csv("outputs/tables/dataset01_vs_dataset02_pivot.csv")

print("\n--- Complete 5-Model Cross-Dataset Comparison Table ---")
display(df_summary)
print("\n--- Side-by-Side Pivot Comparison (Dataset01 vs Dataset02) ---")
display(piv_summary)
```

```python
# ==============================================================================
# STEP 3B: PUBLICATION CHARTS, ROC/PR CURVES, CONFUSION MATRIX, SPATIAL MAP & ONNX
# ==============================================================================
plt.style.use("seaborn-v0_8-whitegrid")

# ------------------------------------------------------------------------------
# FIGURE 1: Training & Validation Convergence + Multi-Horizon MAE Comparison
# ------------------------------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Top-Left: Training & Validation Huber Loss Curves on Dataset 02
ds2_hist = training_histories["Dataset02_MultiModal_SpatialTopology"]
colors = ["#64748b", "#0284c7", "#7c3aed", "#16a34a", "#ea580c"]
for (m_name, h), col in zip(ds2_hist.items(), colors):
    short = m_name.split("_", 1)[1]
    axes[0, 0].plot(range(1, len(h["val_loss"]) + 1), h["val_loss"], "o-", color=col, linewidth=2, label=f"{short} (Val)")
    axes[0, 0].plot(range(1, len(h["train_loss"]) + 1), h["train_loss"], "--", color=col, alpha=0.55)
axes[0, 0].set_title("Validation & Training Huber Loss Convergence (Dataset 02)", fontsize=11, fontweight="bold")
axes[0, 0].set_xlabel("Epoch")
axes[0, 0].set_ylabel("Huber Loss (Z-Score Space)")
axes[0, 0].legend(fontsize=8)

# Top-Right: Overall MAE & RMSE Comparison Across Dataset01 vs Dataset02
x_models = np.arange(5)
w = 0.36
d1_sub = df_summary[df_summary["Dataset"] == "Dataset01_Physical_RoadGraph"]
d2_sub = df_summary[df_summary["Dataset"] == "Dataset02_MultiModal_SpatialTopology"]
axes[0, 1].bar(x_models - w/2, d1_sub["Overall_MAE"], w, label="Dataset 01 (Physical Graph)", color="#3b82f6")
axes[0, 1].bar(x_models + w/2, d2_sub["Overall_MAE"], w, label="Dataset 02 (Multi-Modal Topology)", color="#10b981")
axes[0, 1].set_xticks(x_models)
axes[0, 1].set_xticklabels([m.split("_", 1)[1] for m in d2_sub["Model"]], rotation=15, ha="right", fontsize=9)
axes[0, 1].set_title("Cross-Dataset Overall MAE Comparison (Lower is Better)", fontsize=11, fontweight="bold")
axes[0, 1].set_ylabel("Overall MAE (mph)")
axes[0, 1].legend(fontsize=9)

# Bottom-Left & Bottom-Right: 15-min, 30-min, 60-min Horizon Scaling
for ax, (ds_label, sub_df) in zip([axes[1, 0], axes[1, 1]], [("Dataset 01: Physical Road Graph", d1_sub), ("Dataset 02: Multi-Modal Spatial Topology", d2_sub)]):
    xh = np.arange(3)
    bw = 0.15
    for idx, row in enumerate(sub_df.itertuples(index=False)):
        vals = [row[7], row[8], row[9]]  # 15min_MAE, 30min_MAE, 60min_MAE
        ax.bar(xh + (idx - 2) * bw, vals, bw, label=row.Model.split("_", 1)[1], color=colors[idx])
    ax.set_title(f"Multi-Horizon Error Scaling ({ds_label})", fontsize=11, fontweight="bold")
    ax.set_xticks(xh)
    ax.set_xticklabels(["15-Min Ahead", "30-Min Ahead", "60-Min Ahead"])
    ax.set_ylabel("MAE (mph)")
    ax.legend(fontsize=8, loc="upper left")

plt.tight_layout()
plt.savefig("outputs/figures/fig1_training_and_multihorizon_benchmark.png", dpi=180)
plt.show()

# ------------------------------------------------------------------------------
# FIGURE 2: ROC Curves, Precision-Recall Curves & Confusion Matrix for Severe Congestion
# ------------------------------------------------------------------------------
y_true_bin = (d2_test_targets[:, 5, :].ravel() < cfg.congestion_threshold_mph).astype(int)  # 30-min horizon

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

roc_rows = []
for (m_name, preds_arr), col in zip(d2_test_preds.items(), colors):
    pred_30m = preds_arr[:, 5, :].ravel()
    cong_prob = 1.0 / (1.0 + np.exp((pred_30m - cfg.congestion_threshold_mph) / 4.0))
    fpr, tpr, _ = roc_curve(y_true_bin, cong_prob)
    roc_auc = auc(fpr, tpr)
    prec, rec, _ = precision_recall_curve(y_true_bin, cong_prob)
    pr_auc = average_precision_score(y_true_bin, cong_prob)
    short = m_name.split("_", 1)[1]
    axes[0].plot(fpr, tpr, color=col, linewidth=2.1, label=f"{short} (AUC = {roc_auc:.4f})")
    axes[1].plot(rec, prec, color=col, linewidth=2.1, label=f"{short} (AP = {pr_auc:.4f})")
    roc_rows.append({"Model": m_name, "ROC_AUC_30min": round(roc_auc, 4), "PR_AUC_30min": round(pr_auc, 4)})

axes[0].plot([0, 1], [0, 1], "k--", linewidth=1, label="Random Classifier (AUC = 0.50)")
axes[0].set_title(f"ROC Curve: 30-Min Congestion Detection (<{cfg.congestion_threshold_mph} mph)", fontsize=11, fontweight="bold")
axes[0].set_xlabel("False Positive Rate (1 - Specificity)")
axes[0].set_ylabel("True Positive Rate (Sensitivity)")
axes[0].legend(fontsize=8, loc="lower right")

axes[1].set_title("Precision-Recall Curve: 30-Min Congestion Alerting", fontsize=11, fontweight="bold")
axes[1].set_xlabel("Recall (Congestion Capture Rate)")
axes[1].set_ylabel("Precision (Alert Reliability)")
axes[1].legend(fontsize=8, loc="lower left")

# Confusion Matrix for Proposed RA-DynSTGCN at 30-Min Horizon
stgcn_pred_bin = (d2_test_preds["4_Proposed_RA_DynSTGCN"][:, 5, :].ravel() < cfg.congestion_threshold_mph).astype(int)
cm = confusion_matrix(y_true_bin, stgcn_pred_bin)
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=axes[2],
    xticklabels=["Free Flow (>=45)", "Congested (<45)"],
    yticklabels=["Free Flow (>=45)", "Congested (<45)"]
)
axes[2].set_title("Confusion Matrix: Proposed RA-DynSTGCN (30-Min)", fontsize=11, fontweight="bold")
axes[2].set_xlabel("Predicted Traffic Regime")
axes[2].set_ylabel("Actual Traffic Regime")

plt.tight_layout()
plt.savefig("outputs/figures/fig2_congestion_roc_pr_confusion.png", dpi=180)
plt.show()

display(pd.DataFrame(roc_rows))

# ------------------------------------------------------------------------------
# FIGURE 3: Dynamic Adjacency Heatmap, Spatial Sensor Map & Multi-Horizon Trajectory
# ------------------------------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

im = axes[0].imshow(d2_dyn_adj[:45, :45], cmap="viridis", aspect="auto")
axes[0].set_title("Learned Regime-Aware Dynamic Graph A_dyn", fontsize=11, fontweight="bold")
axes[0].set_xlabel("Target Sensor Index")
axes[0].set_ylabel("Source Sensor Index")
plt.colorbar(im, ax=axes[0], label="Dynamic Edge Weight")

coords = ds2["coords"]
pred_30m_snapshot = d2_test_preds["4_Proposed_RA_DynSTGCN"][0, 5, :]
sc = axes[1].scatter(coords[:, 1], coords[:, 0], c=pred_30m_snapshot, cmap="RdYlGn", vmin=20, vmax=68, s=42, edgecolors="k", linewidths=0.35)
axes[1].set_title("LA Highway Network: 30-Min Forecast (mph)", fontsize=11, fontweight="bold")
axes[1].set_xlabel("Longitude")
axes[1].set_ylabel("Latitude")
plt.colorbar(sc, ax=axes[1], label="Predicted Speed (mph)")

steps_min = np.arange(1, 13) * 5
sensor_idx = 12
axes[2].plot(steps_min, d2_test_targets[0, :, sensor_idx], "ko-", linewidth=2.3, label="Ground Truth")
for (m_name, preds_arr), col in zip(d2_test_preds.items(), colors):
    axes[2].plot(steps_min, preds_arr[0, :, sensor_idx], "--", color=col, linewidth=1.7, label=m_name.split("_", 1)[1])
axes[2].set_title(f"60-Min Multi-Horizon Forecast Trajectory (Sensor #{sensor_idx})", fontsize=11, fontweight="bold")
axes[2].set_xlabel("Forecast Horizon (Minutes Ahead)")
axes[2].set_ylabel("Traffic Speed (mph)")
axes[2].legend(fontsize=7.5)

plt.tight_layout()
plt.savefig("outputs/figures/fig3_dynamic_graph_spatial_trajectory.png", dpi=180)
plt.show()

# ------------------------------------------------------------------------------
# STEP 3C: INTERACTIVE SPATIAL HEATMAP & ONNX EDGE LATENCY BENCHMARK
# ------------------------------------------------------------------------------
def export_interactive_map_and_onnx(model_cpu, sample_inp, coords, sensor_ids, pred_speeds):
    html_out = "outputs/figures/la_30min_congestion_map.html"
    try:
        import plotly.express as px
        df_map = pd.DataFrame({
            "sensor_id": sensor_ids,
            "latitude": coords[:, 0],
            "longitude": coords[:, 1],
            "predicted_speed_mph": np.round(pred_speeds, 2),
            "regime": np.where(pred_speeds < 35, "Severe Congestion",
                      np.where(pred_speeds < 50, "Moderate Traffic", "Free Flow")),
        })
        fig = px.scatter_mapbox(
            df_map, lat="latitude", lon="longitude", color="predicted_speed_mph",
            size=np.clip(70.0 - df_map["predicted_speed_mph"], 8, 35),
            hover_name="sensor_id", hover_data=["regime", "predicted_speed_mph"],
            color_continuous_scale="RdYlGn", range_color=[15, 70], zoom=9.5,
            mapbox_style="carto-positron",
            title="LA Urban Network 30-Min Predicted Speed & Congestion Heatmap (RA-DynSTGCN)"
        )
        fig.write_html(html_out)
    except Exception as e:
        print("Plotly HTML export note:", e)

    onnx_path = "outputs/onnx/ra_dynstgcn_d2.onnx"
    model_cpu.eval()
    x1 = sample_inp[:1].cpu()
    onnx_ok = False
    try:
        import onnxscript  # Required by PyTorch 2.x ONNX exporter
    except ImportError:
        os.system("pip install -q onnxscript onnxruntime")
    try:
        torch.onnx.export(
            model_cpu, x1, onnx_path,
            input_names=["traffic_window"], output_names=["multi_horizon_speed"],
            dynamic_axes={"traffic_window": {0: "batch"}, "multi_horizon_speed": {0: "batch"}},
            opset_version=14
        )
        onnx_ok = os.path.exists(onnx_path)
    except Exception as e:
        print("ONNX export note:", e)

    with torch.no_grad():
        for _ in range(15):
            _ = model_cpu(x1)
        t0 = time.perf_counter()
        for _ in range(100):
            _ = model_cpu(x1)
        pt_latency_ms = (time.perf_counter() - t0) / 100.0 * 1000.0

    profile = {
        "onnx_exported": onnx_ok,
        "onnx_size_kb": round(os.path.getsize(onnx_path) / 1024.0, 2) if onnx_ok else 0.0,
        "pytorch_single_frame_latency_ms": round(pt_latency_ms, 3),
        "sub_second_idas_compliant": pt_latency_ms < 1000.0,
    }
    try:
        if onnx_ok:
            import onnxruntime as ort
            sess = ort.InferenceSession(onnx_path, providers=["CPUExecutionProvider"])
            x1_np = x1.numpy()
            for _ in range(15):
                sess.run(None, {"traffic_window": x1_np})
            t0 = time.perf_counter()
            for _ in range(100):
                sess.run(None, {"traffic_window": x1_np})
            profile["onnxruntime_single_frame_latency_ms"] = round((time.perf_counter() - t0) / 100.0 * 1000.0, 3)
    except Exception as e:
        profile["onnxruntime_note"] = str(e)

    with open("outputs/onnx/edge_latency_profile.json", "w") as f:
        json.dump(profile, f, indent=2)
    return profile

edge_profile = export_interactive_map_and_onnx(
    trained_d2_stgcn, d2_sample_x, ds2["coords"], ds2["sensor_ids"],
    d2_test_preds["4_Proposed_RA_DynSTGCN"][0, 5, :]
)
print("=== Edge Inference & ONNX Deployment Profile ===")
print(json.dumps(edge_profile, indent=2))
```
